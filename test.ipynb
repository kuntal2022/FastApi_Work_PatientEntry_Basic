```python
from pydantic import BaseModel, model_validator, field_validator, computed_field
from typing import *
```

```python
class Pat(BaseModel):
    name : str 
    age : int

dict_x = {"name":"ram", "age":20 }
pat1=Pat(**dict_x)

pat1
```

```python
class Address(BaseModel):
    city: str 
    state: str 
    house_no: int 
    pin: int 



class Person(BaseModel):
    name: str 
    age: int 
    weight: float
    height: float
    address: Address


    @computed_field
    @property
    def height_in_m(self)->float:
        return round((self.height * 0.3048),2)

    @computed_field
    @property
    def bmi(self)->float: 
        return round(self.weight/(self.height_in_m)**2,2)

 

        
    



address_dict ={"city": "Bangalore", 'house_no': 19, 'state': "Karnataka",  "pin":712405}
address1=Address(**address_dict)


person_dict ={"name": "Prem",  'height':5.2, "weight": 75.0, "age": 30, "address":address1 }

person1=Person(**person_dict)
temp1=person1.model_dump()
temp2=person1.model_dump_json()
```
